```python
import sys
sys.path.append("..")

import plotly.graph_objects as go
from ramancore.io import read_spectrum
from ramancore.preprocess import crop, remove_spikes
```

## 1. Демо-спектр с искусственными космическими лучами

```python
x, y, meta = read_spectrum("../data/examples/synthetic_si.txt")

y_spiky = y.copy()      # .copy() — делаем копию, чтобы не испортить исходный y
y_spiky[300] += 2500     # луч в 1 точку
y_spiky[1200] += 1800   # луч в 2 точки
y_spiky[1201] += 2200

y_clean = remove_spikes(y_spiky)

fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y_spiky, name="с лучами"))
fig.add_trace(go.Scatter(x=x, y=y_clean, name="после remove_spikes"))
fig.update_layout(xaxis_title="Рамановский сдвиг, см⁻¹", yaxis_title="Интенсивность")
fig.show()
```

Лучи должны исчезнуть, а пик Si у 520.7 см⁻¹ — остаться без изменений.
