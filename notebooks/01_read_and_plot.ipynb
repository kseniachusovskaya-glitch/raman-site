Запускайте ячейки по очереди: Shift+Enter.

```python
# Импорт: подключаем нашу функцию из ядра и библиотеку графиков plotly
import sys
sys.path.append("..")  # чтобы Python увидел папку ramancore на уровень выше

import plotly.graph_objects as go
from ramancore.io import read_spectrum
```

## 1. Читаем демо-спектр

```python
x, y, meta = read_spectrum("../data/examples/synthetic_si.txt", laser_nm=532)
print(meta)
print("Точек:", len(x))
```

```python
# Строим интерактивный график: колесо мыши — масштаб, двойной щелчок — сброс
fig = go.Figure(go.Scatter(x=x, y=y, mode="lines", name=meta["filename"]))
fig.update_layout(xaxis_title="Рамановский сдвиг, см⁻¹", yaxis_title="Интенсивность")
fig.show()
```
