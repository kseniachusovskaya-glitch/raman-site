```python
import sys
sys.path.append("..")

import numpy as np
import plotly.graph_objects as go
from ramancore.io import read_spectrum
from ramancore.preprocess import crop, remove_spikes, smooth
```

## Ваш спектр с разными окнами

Окно задаётся в **точках**. Посмотрите, при каком окне шум уже убран, а пики ещё не стали ниже и шире.

```python
my_file = "../data/examples/synthetic_si.txt"  # ← подставьте свой файл
x, y, meta = read_spectrum(my_file)
x, y = crop(x, y, 100, 3200)
y = remove_spikes(y)

print("Шаг по x, см⁻¹:", np.median(np.diff(x)))

fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y, name="без сглаживания"))
for window in [5, 11, 21]:   # можно поменять список окон
    ys = smooth(y, window=window)
    fig.add_trace(go.Scatter(x=x, y=ys, name=f"окно {window}"))
    print(f"окно {window}: максимум {ys.max():.0f} при x = {x[np.argmax(ys)]:.1f}")
fig.update_layout(xaxis_title="Рамановский сдвиг, см⁻¹", yaxis_title="Интенсивность")
fig.show()
```
