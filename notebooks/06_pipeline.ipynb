# Этап 1: весь конвейер обработки

`process` делает по порядку: crop → remove_spikes → subtract_baseline → (smooth) → resample → normalize. Так же будут обрабатываться эталоны базы.

Нажмите **Run All**, затем подставьте свой файл.

```python
import sys
sys.path.append("..")

import plotly.graph_objects as go
from ramancore.io import read_spectrum
from ramancore.preprocess import process
```

```python
my_file = "../data/examples/synthetic_si.txt"  # ← подставьте свой файл
x, y, meta = read_spectrum(my_file, laser_nm=532)

grid, yp, meta = process(x, y, meta)   # все параметры по умолчанию

for step in meta["history"]:   # что и с какими параметрами сделано
    print(step)
```

```python
fig = go.Figure(go.Scatter(x=grid, y=yp, name="после конвейера"))
fig.update_layout(xaxis_title="Рамановский сдвиг, см⁻¹", yaxis_title="Нормированная интенсивность")
fig.show()
```
